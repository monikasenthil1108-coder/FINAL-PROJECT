**Dataset Column/features Description:**

* **City**- Name of the Indian city where the air quality was measured

* **Date** - The date on which the readings were recorded

* **PM2.5** - Concentration of fine particulate matter (tiny dust/smoke particles) in the air

* **PM10** - Concentration of coarser particulate matter in the air

* **NO2** - Nitrogen Dioxide level, mainly from vehicles and industry

* **NOx** - Total nitrogen oxides in the air

* **NH3** - Ammonia level, mainly from vehicles and industry

* **CO** - Carbon Monoxide level, from burning fuel/vehicles

* **SO2** - Sulfur Dioxide level, mainly from burning coal/diesel

* **O3** - Ground-level Ozone concentration

* **Benzene** - A chemical pollutant from vehicle fuel and industry

* **Toluene** - A chemical pollutant, similar source as Benzene

* **Xylene** - A chemical pollutant, similar source as Benzene

* **AQI** - Air Quality Index, a single combined score representing overall air pollution level

* **AQI_Bucket** - Category label based on AQI score (Good, Satisfactory, Moderate, Poor, Very Poor, Severe)

2.2 Correcting Data Types

```python
#Conver 'Date' column from text (object) to proper datatime format
df['Date']=pd.to_datetime(df['Date'])

# Confirm the change
print(df['Date'].dtype)
df.head()
```

2.3 Handling Missing Values


```python
# Removing Xylene column since 61% of its values are missing--too unreliable to fix
df.drop(columns=['Xylene'], inplace=True)
```

```python
#Filling missing pollutant values using each city's median
pollutant_cols = ['PM2.5', 'PM10', 'NO', 'NO2','NOx', 'NH3', 'CO', 'SO2', 'O3',
                 'Benzene', 'Toluene' ]
```

```python
for col in pollutant_cols:
   df[col] = df.groupby('City')[col]. transform(lambda x: x.fillna(x.median()))
```

```python
#Filling any leftover gaps(where a city had zero data for a pollutant) using the overall median
for col in pollutant_cols:
  df[col] = df[col].fillna(df[col].median())
```

```python
# removing rows where AQI is missing, since it's our key metric and too risky to guess.
df.dropna(subset=['AQI', 'AQI_Bucket'], inplace=True)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
print("\nNew shape:", df.shape)
```

2.4 Creating Derived Columns

```python
#Extract Year and Month from Date
df['Year'] = df['Date']. dt.year
df['Month'] = df['Date']. dt.month
df['Month_Name'] = df['Date']. dt.month_name()
```

```python
#Create a Season column based on Indian Seasonal Patterns
def get_season(month):
  if month in [12, 1, 2]:
    return 'Winter'
  elif month in [3, 4, 5]:
    return 'Summer'
  elif month in [6, 7, 8, 9]:
    return 'Monsoon'
  else:
    return 'Post-Monsoon'

df['Season'] = df['Month']. apply(get_season)

df[['Date', 'Year', 'Month', 'Month_Name', 'Season']].head()
```

2.6 Checking Skewness of Numerical Columns

```python
numeric_cols = pollutant_cols + ['AQI']
```

```python
print("skewness of numerical columns:")
for col in numeric_cols:
  print(f"{col}:{df[col].skew():.2f}")
```

**Observation**: All numeric columns show positive(right) skewness, with O3 being the mildest(1.35) and Benzene the most extreme(23.25). Columns like CO, Toluene, and Benzene show unusally high skewness, suggesting a small number of extreme values are heavily pulling their distributions. This will be investigated further in the outlier detection step below.

2.7 Checking Outliers (IQR Method)

```python
def detect_outliers_iqr(column):
  Q1 = df[column].quantile(0.25)
  Q3 = df[column].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
  return len(outliers), lower_bound, upper_bound
```

```python
print("Outliers check(IQR method):")
for col in numeric_cols:
  count, lower, upper = detect_outliers_iqr(col)
  pct = round(count / len(df) * 100, 2)
  print(f"{col}: {count} outliers ({pct}%) | Normal range: {lower:.1f} to {upper:.1f}")
```

**Observation:** Between 3% and 12% of values in each column fall outside the normal IQR range, with Toluene having the most (12.34%) and O3 the least (2.97%). Note that the lower bounds for most columns are negative (e.g PM2.5: -45.7) -- since pollution values can never be negative in real life, this means there are effectively no meaningful "too low" outliers here, only high- side ones matter.


```python
for col in numeric_cols:
  print(f"{col}: min{df[col].min()}, max={df[col].max()}")
```

Observation: Minimum values across all columns are close to 0, which is expected for pollution data. Maximum values for PM2.5 (914.94) and PM10 (917.08) are extreme but plausible, matching real severe pollution events recorded in Indian cities during winter smog. The AQI maximum of 2049 is notable since it exceeds the standard 0-500 AQI scale - this is flagged as an unusual value, but not removed, since we cannot confirm with certainty whether it reflects a data reporting choice or a genuine extreme case.


**Decision: Outliers are kept, not removed.** In air quality data, extreme values usually represent real, meaningful pollution events-- such as Diwali fireworks, stubble burning season, or severe winter smog -- rather than data entry errors. Removing them would high genuinely important pollution spikes that are central to this analysis.

# Analysis 1: Most Polluted Cities in India(Bivariate Analysis)

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style('whitegrid')

top10_cities = df.groupby('City')['AQI'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top10_cities.values,  y=top10_cities.index, hue=top10_cities.index,
            palette='Reds_r', legend=False)
plt.title('Top 10 Most Polluted Cities in India',
          fontsize=14, fontweight='bold')
plt.xlabel('Average AQI')
plt.ylabel('City')
plt.tight_layout()
plt.show()
```

**Insights**


* Ahmedabad and Delhi show the highest average AQI among all cities, indicating consistently poor air quality.   
* Cities in the top 10 likely share common pollution sources such as heavy traffic, industrial activity, or seasonal crop burning nearby.
* Cities outside the top 10 have comparatively better air quality on average, but may still experience seasonal spikes.
